```python
import os
import glob
import matplotlib.pyplot as plt
import numpy as np
import json

DATA_PATH = r"path\to\data"

# Step 1: Get genre distribution
files_by_genre = {}
for genre in os.listdir(os.path.join(DATA_PATH, "train")):
    TEST_PATH = os.path.join(DATA_PATH, "test", genre)
    TRAIN_PATH = os.path.join(DATA_PATH, "train", genre)
    
    test_files = glob.glob(f"{TEST_PATH}/**/*.txt", recursive=True)
    train_files = glob.glob(f"{TRAIN_PATH}/**/*.txt", recursive=True)
    
    # Add counts to dictionary
    files_by_genre[genre] = {
        "test": len(test_files),
        "train": len(train_files),
        "total": len(test_files) + len(train_files)
    }

# Sort genres by the total number of files (descending order)
sorted_genres = sorted(files_by_genre.keys(), key=lambda g: files_by_genre[g]["total"], reverse=True)

# Step 2: Load losses for each genre
losses_by_genre = {}
for genre in sorted_genres:
    losses_path = os.path.join(DATA_PATH, "models", genre, "losses.json")
    if os.path.exists(losses_path):
        with open(losses_path, "r") as f:
            data = json.load(f)
        losses_by_genre[genre] = data

# Step 3: Define consistent colors for genres
color_cycle = plt.cm.tab10.colors  # Use a colormap with distinct colors
genre_colors = {genre: color_cycle[i % len(color_cycle)] for i, genre in enumerate(sorted_genres)}

# Step 4: Plot maximum 6 genres per plot and save them
chunk_size = 6
chunks = [sorted_genres[i:i + chunk_size] for i in range(0, len(sorted_genres), chunk_size)]

output_dir = os.path.join(DATA_PATH, "plots")
os.makedirs(output_dir, exist_ok=True)

for i, chunk in enumerate(chunks):
    plt.figure(figsize=(12, 8))
    for genre in chunk:
        if genre in losses_by_genre:
            train_losses = losses_by_genre[genre]["train_losses"]
            test_losses = losses_by_genre[genre]["test_losses"]
            epochs = np.arange(1, len(train_losses) + 1)

            # Use the same color for train and test lines of the same genre
            color = genre_colors[genre]
            plt.plot(epochs, train_losses, label=f"{genre} - Train", linestyle="--", color=color)
            plt.plot(epochs, test_losses, label=f"{genre} - Test", linestyle="-", color=color)

    # Customize the plot
    plt.title(f"Losses for Genres: {', '.join(chunk)}")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend(loc="upper right", fontsize="small", ncol=2)
    plt.grid()
    plt.tight_layout()

    # Save plot to PNG
    output_path = os.path.join(output_dir, f"loss_plot_{i+1}.png")
    plt.savefig(output_path)
    plt.close()

```
