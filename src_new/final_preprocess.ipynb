```python
import glob
import os
import random
import shutil
import json
from midi import generate_word_files
from model_without_genre import build_vocab

before_path = r"C:\Users\WDolibóg\Desktop\adl-piano-midi"
data_path = r"C:\projects\studia\POLSLrepo_sem7\music_generator\data\all_separately2"
train_path = os.path.join(data_path, "train")
test_path = os.path.join(data_path, "test")
models_path = os.path.join(data_path, "models")

os.makedirs(train_path, exist_ok=True)
os.makedirs(test_path, exist_ok=True)
os.makedirs(models_path, exist_ok=True)
```

```python
files_num = dict()

for root, dirs, files in os.walk(before_path):
   for name in files:
       ext = os.path.splitext(name)[1]
       if ext in files_num:
           files_num[ext]+=1
       else:
           files_num[ext] = 1
print(files_num)
for m in  glob.glob(os.path.join(data_path,"/**/*.mid"), recursive=True):
    os.remove(m)
```

```python
all = [f for f in glob.glob(f"{before_path}/**/*.mid", recursive=True)]

print(len(all))
random.shuffle(all)

test_len = int(0.15 * len(all))
train_len = len(all) - test_len

print(f"test_len == {test_len}, train_len == {train_len}")
test = all[:test_len]
train = all[test_len:]

print(len(test))
print(len(train))


for f in test:
    relative_path = os.path.relpath(f, before_path)
    dest = os.path.join(test_path, relative_path)
    os.makedirs(os.path.split(dest)[0], exist_ok=True)
    try:
        shutil.copy(f, dest)
    except Exception as e:
        print(f"{f} ---- {e}")

for f in train:
    relative_path = os.path.relpath(f, before_path)
    dest = os.path.join(train_path, relative_path)
    os.makedirs(os.path.split(dest)[0], exist_ok=True)
    try:
        shutil.copy(f, dest)
    except Exception as e:
        print(f"{f} ---- {e}")
```
