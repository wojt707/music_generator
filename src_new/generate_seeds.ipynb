```python
import os
import random
import glob
import json


def create_genre_seeds(
    test_folder: str, output_folder: str, word_to_idx: dict, seeds_per_genre: int = 100
):
    """
    Generate seed sequences for each genre based on the test dataset.

    Args:
    - test_folder (str): Path to the test dataset, organized into genre folders.
    - output_folder (str): Path to save the generated `seeds.txt` for each genre.
    - word_to_idx (dict): Mapping from words to their corresponding indices.
    - seeds_per_genre (int): Number of seed sequences to generate per genre.
    """
    os.makedirs(output_folder, exist_ok=True)

    for genre in os.listdir(test_folder):
        print(genre)

        genre_path = os.path.join(test_folder, genre)
        if not os.path.isdir(genre_path):
            continue

        # Collect all word sequences from `.txt` files in the genre folder
        token_sequences = []
        for txt in glob.glob(f"{genre_path}/**/*.txt", recursive=True):
            with open(txt, "r") as f:
                words = f.read().split()
                filtered = [str(word_to_idx[word]) for word in words if word != "PAD"]
                token_sequences.extend(filtered)
        print(f"token_len = {len(token_sequences)}")
        # Generate random seed sequences
        seeds = []
        for _ in range(seeds_per_genre):
            seed_seq_length = random.randint(10, 32)
            start_idx = random.randint(
                0, max(0, len(token_sequences) - seed_seq_length - 1)
            )
            seed = token_sequences[start_idx : start_idx + seed_seq_length]
            seeds.append(" ".join(seed))

        output_file = os.path.join(output_folder, f"{genre.lower()}_seeds.txt")
        with open(output_file, "w") as f:
            f.write("\n".join(seeds))
        print(f"Generated {len(seeds)} seeds for genre '{genre.lower()}'.")


# Example usage
test_folder = r"path\to\data\test"
output_folder = r"path\to\data\seeds"
with open(r"path\to\word_to_idx.json", "r") as f:
    word_to_idx = json.load(f)
create_genre_seeds(test_folder, output_folder, word_to_idx, seeds_per_genre=100)
```
