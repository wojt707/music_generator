```python
import glob
import os
import matplotlib.pyplot as plt
import numpy as np

DATA_PATH = r"path\to\data"

files_by_genre = {}

for genre in os.listdir(os.path.join(DATA_PATH, "train")):
    TEST_PATH = os.path.join(DATA_PATH, "test", genre)
    TRAIN_PATH = os.path.join(DATA_PATH, "train", genre)
    
    test_files = glob.glob(f"{TEST_PATH}/**/*.txt", recursive=True)
    train_files = glob.glob(f"{TRAIN_PATH}/**/*.txt", recursive=True)
    
    # Add counts to dictionary
    files_by_genre[genre] = {
        "test": len(test_files),
        "train": len(train_files)
    }
print(files_by_genre)
# Plotting
genres = list(files_by_genre.keys())
train_counts = [files_by_genre[genre]["train"] for genre in genres]
test_counts = [files_by_genre[genre]["test"] for genre in genres]

x = np.arange(len(genres))  # Create positions for the bars
width = 0.35  # Width of the bars

# Create the plot
plt.figure(figsize=(10, 6))
plt.bar(x - width / 2, train_counts, width, label="Train Files", color="blue", alpha=0.7)
plt.bar(x + width / 2, test_counts, width, label="Test Files", color="orange", alpha=0.7)

# Add labels and title
plt.xlabel("Genre")
plt.ylabel("Number of Files")
# plt.title("Number of Train and Test Files by Genre")
plt.xticks(x, genres, rotation=45)
plt.legend()
plt.tight_layout()

# Save the plot
output_path = os.path.join(DATA_PATH, "file_distribution_separate.png")
plt.savefig(output_path)
print(f"Plot saved to {output_path}")
    
```

```python
files_num = dict()

for root, dirs, files in os.walk(DATA_PATH):
    for name in files:
        ext = os.path.splitext(name)[1]
        if ext in files_num:
            files_num[ext] += 1
        else:
            files_num[ext] = 1
files_num
```
